```python
rows = []
file = "C:\\Users\\zuzka\\Desktop\\Sample_SVU_60nm_slot-2\\Section_1_Polygon_pt3.mrc.mdoc"
with open(file, mode = "r") as s:
    text = s.readlines()
    a=0
    for line in text:
        if line.startswith("PieceCoordinates"):
            splits = line.split(" ")
            #print(x[2], x[3])
            coordx,coordy = str(splits[2]), str(splits[3])
            row = "img"+str("{0:03}".format(a))+".tif; ;("+ coordx + ", " + coordy +")" 
            #print(row)
            rows.append(row)
            a +=1

print(rows)
```

```python
output = "C:\\Users\\zuzka\\Desktop\\Sample_SVU_60nm_slot-2\\configfile.txt"
with open(output, mode = "w") as sw:
    sw.write("dim = 2")
    sw.write("\n")
    for l in rows:
        sw.write(l)
        sw.write("\n")
```
