```python
import os
import shutil

# vybrat slozku Points (obsahujici obrazky z makra koncici spots.tif)
src = r'D:\Users data\Margarita\2023-02-14-16_Points'
# zalozit slozku Spots
dest = r'D:\Users data\Margarita\2023-02-14-16_Spots' 

for path, subdirs, files in os.walk(src): 
    for name in files:
        if name.endswith("_spots.tif"):
            filename = os.path.join(path, name)
            shutil.copy2(filename, dest)
```
