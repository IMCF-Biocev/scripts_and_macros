```python
import os
import shutil

src = r'D:\Users data\Margarita\2023-02-14-16_Images'
dest = r'D:\Users data\Margarita\2023-02-14-16_Masks'

for path, subdirs, files in os.walk(src): 
    for name in files:
        if name.endswith("_mask.tif"):
            filename = os.path.join(path, name)
            print(filename)
            shutil.copy2(filename, dest)
```

```python
ext = ".tif_mask.tif"
for filen in os.listdir(dest):
        if ext in filen:
            print(filen)
            x = filen
            x = x.replace(".tif_mask.tif", "_mask.tif")
            print(x)
            os.rename(os.path.join(dest,filen), os.path.join(dest,x))
```
